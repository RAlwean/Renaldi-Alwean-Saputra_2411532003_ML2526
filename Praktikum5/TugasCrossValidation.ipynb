```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold, cross_val_score
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import LabelEncoder

df = pd.read_csv('titanic_data/train.csv')
print("5 Data Teratas:")
df.head()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Embarked'] = df['Embarked'].fillna(df['Embarked'].mode()[0])

df = df.drop(['Name', 'Ticket', 'Cabin', 'PassengerId'], axis=1)

le = LabelEncoder()
df['Sex'] = le.fit_transform(df['Sex'])
df['Embarked'] = le.fit_transform(df['Embarked'])

X_titanic = df.drop('Survived', axis=1)
y_titanic = df['Survived']

print("Data setelah praproses:")
X_titanic.head()
```

```python
model_titanic = LogisticRegression(max_iter=1000)

kf_titanic = KFold(n_splits=5, shuffle=True, random_state=42)

acc_scores = cross_val_score(model_titanic, X_titanic, y_titanic, cv=kf_titanic, scoring='accuracy')

print("=== Analisis Kinerja Model (Titanic) ===")
print(f"Akurasi tiap fold: {acc_scores}")
print(f"Rata-rata Akurasi: {acc_scores.mean():.4f}")
print(f"Standar Deviasi (Variasi): {acc_scores.std():.4f}")
```
