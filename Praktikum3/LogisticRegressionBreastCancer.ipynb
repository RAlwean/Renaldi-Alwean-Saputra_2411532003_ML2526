### Memisahkan Fitur dan Variabel Target

Kita akan memisahkan dataset menjadi fitur (X) dan variabel target (y).
Dalam kasus ini, 'diagnosis' adalah variabel target.

```python
import pandas as pd

#load dataset
dataset_url = 'https://raw.githubusercontent.com/RAlwean/Renaldi-Alwean-Saputra_2411532003_ML2526/refs/heads/main/Praktikum2/Breast_cancer_data.csv'
df = pd.read_csv(dataset_url)

X = df.drop('diagnosis', axis=1)
y = df['diagnosis']

print("Bentuk X:", X.shape)
print("Bentuk y:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=16, stratify=y)

print("Bentuk X_train:", X_train.shape)
print("Bentuk X_test:", X_test.shape)
print("Bentuk y_train:", y_train.shape)
print("Bentuk y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score

# 1. Inisialisasi model baru untuk data tanpa scaling
logreg_no_scaling = LogisticRegression(random_state=16, max_iter=1000)

# 2. Latih model menggunakan data asli (X_train), BUKAN X_train_scaled
logreg_no_scaling.fit(X_train, y_train)

# 3. Lakukan prediksi menggunakan data asli (X_test)
y_pred_no_scaling = logreg_no_scaling.predict(X_test)

# 4. Evaluasi hasil tanpa scaling
print("--- Kinerja Model TANPA Scaling ---")
print(f"Accuracy: {accuracy_score(y_test, y_pred_no_scaling):.4f}")
print(classification_report(y_test, y_pred_no_scaling))
```
